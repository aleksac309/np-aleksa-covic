```python
rezervacije = [
    {"gost": "Ana", "noci": 3, "cijena_noci": 80},
    {"gost": "Marko", "noci": 5, "cijena_noci": 65},
    {"gost": "Ivana", "noci": 2, "cijena_noci": 120},
]

# Eksplicitna petlja
ukupno = 0
for r in rezervacije:
    ukupno += r["noci"] * r["cijena_noci"]

# Ekvivalent pomoću izraza
iznosi = [r["noci"] * r["cijena_noci"] for r in rezervacije]
skupa_nocenja = [r["gost"] for r, i in zip(rezervacije, iznosi) if i >= 240]

print(ukupno, sum(iznosi))   # 805 805
print(skupa_nocenja)                 # ['Marko']

```

```python
def iznos(noci: int, cijena_noci: float,
          popust: float = 0.0) -> float:
    """Iznos rezervacije umanjen za popust (0.0 do 1.0)."""
    if not 0.0 <= popust <= 1.0:
        raise ValueError("Popust mora biti u intervalu [0, 1]")
    return noci * cijena_noci * (1 - popust)

print(iznos(3, 80))           # 240.0
print(iznos(3, 80, 0.1))      # 216.0

try:
    iznos(3, 80, 1.5)
except ValueError as e:
    print("Greška:", e)

```

```python
import csv
from pathlib import Path
def prosjecna_cijena(putanja: Path) -> float:
    ukupno, noci = 0.0, 0
    with putanja.open(newline="", encoding="utf-8") as f:
        for red in csv.DictReader(f):
            n = int(red["noci"])
            ukupno += n * float(red["cijena_noci"])
            noci += n
    if noci == 0:
        raise ValueError("Nema ostvarenih noćenja")
    return ukupno / noci

```

```python
from pathlib import Path

def test_prosjek():
    p = Path("/content/r.csv")
    p.write_text("gost,noci,cijena_noci\nAna,2,100\nMarko,2,50\n")

    rezultat = prosjecna_cijena(p)

    assert rezultat == 75.0
    print("Test uspješan! Prosječna cijena:", rezultat)
```

```python
from abc import ABC, abstractmethod

class Tarifa(ABC):
    @abstractmethod
    def cijena(self, osnovna: float) -> float: ...

class Standardna(Tarifa):
    def cijena(self, osnovna: float) -> float:
        return osnovna

class Sezonska(Tarifa):
    def cijena(self, osnovna: float) -> float:
        return osnovna * 1.3

def obracun(noci: int, osnovna: float, tarifa: Tarifa) -> float:
    return noci * tarifa.cijena(osnovna)

print(obracun(3, 100, Sezonska()))   # 390.0
print(obracun(3, 100, Standardna()))   # 390.0
```
